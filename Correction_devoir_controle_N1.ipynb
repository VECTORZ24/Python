```python
##Correction du devoir de controle N 1:

#Saisir 4<=n<=12
def Saisie():
    N=int(input("Donner 4<=N<=12: "))
    while not(4<=N<=12):
        N=int(input("Donner 4<=N<=12: "))
    return N
#Remplir un talbeau T de N noms non vide et majuscules:
def RemplirT(T,N):
    for i in range(N):
        T[i]=input("Donner un nom majuscule et non vide: ")
        while not(T[i]!="" and T[i].isupper()):
            T[i]=input("Donner un nom majuscule et non vide: ")
#Sommer les valeurs ord de chaque nom dans T
def Somme(Ch):
    S=0
    for i in Ch:
        S+=ord(i)
    return S
#Remplir le tableau V
def RemplirV(T,V):
    n=len(T)
    for i in range(n):
        V[i]=Somme(T[i])
    return V
#Identifier les nombres premiers
def Premier(N):
    P=True
    for i in range(2,N//2):
        if(N%i==0):
            P=False
    return P
#Trier les noms gagnants
def Tri(Ch):
    Ch=list(Ch) #transformer la chaine en un tableau pour permuter ses caractères
    echange=True
    while echange==True:
        echange=False
        for i in range(len(Ch)-1):
            if(ord(Ch[i])>ord(Ch[i+1])):
                Ch[i+1],Ch[i]=Ch[i],Ch[i+1]
                echange=True
    Ch="".join(Ch) #rendre la liste en chaine des caractères
    return Ch
#Afficher les messages des noms gagnants non triés et triés 
def Afficher(T,V):
    gagnants=""
    gagnants_triés=""
    for i in range(len(V)):
        if(Premier(V[i])==True):
            gagnants = gagnants+T[i]+"  "
            gagnants_triés=gagnants_triés+Tri(T[i])+"  "
    print("les chaines dont leur somme ASCCI est un nombre premier:")
    print(gagnants)
    print("Après trie, les chaines gagnants sont:")
    print(gagnants_triés)
#programme principal
N=Saisie()
T=[""]*N
RemplirT(T,N)
V=[""]*N
RemplirV(T,V)
print(V)
Afficher(T,V)
```
