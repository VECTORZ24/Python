```python
#importing some libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import date
from sklearn.linear_model import LinearRegression
sns.set()
```

```python
data=pd.read_csv("patients.csv")
data
```

```python
#transforming STR to datetime variables
data["departure_date"] = pd.to_datetime(data["departure_date"])
data["arrival_date"] = pd.to_datetime(data["arrival_date"])
data["Stay_period"]=(data["departure_date"]-data["arrival_date"]).dt.days
```

```python
#building a linear regression model
Y=data["satisfaction"]
X=data[["Stay_period"]]
model=LinearRegression()
model.fit(X,Y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
#drawing the linear regression curve (as we can see, the data cannot follow a linear regression model)
yhat=model.coef_*X+model.intercept_
plt.plot(yhat,c="orange",lw=1)
```
